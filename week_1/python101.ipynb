### Data Types

There are about **5** different types of data in python that you need to learn.


1. Strings - Ex: "Name", "Colombo3", "random@gmail.com"
2. Integer - Ex: 12232, 4, -9092332
3. Float - Ex: 3.14129, 1.00, -0.002

4. Lists - Ex: 
    ```python 
    ["Apple", "Banana", "Green"], ["Kothu", "Naan", 9001]
    ```
5. Tuples - Ex:
    ```python
    ("Nizar",26)
    ```

5. Dictionary- Ex: 

    ```python 
    { 
        "Name": "Nizar",
        "age": 26,
        "fav_movies": ["evil dead", "up", "avengers"],
        "height(cm)": 183.09 
    }
    ```


#### 1. Strings

Restrictions:
 - **Double ("" "") or Single (" ") quotes**

````python
# Let's create a variable called name and assign it the value of your name

name = "Your Name" # Replace "Your Name" with your actual name
````

````python
# Print the value of the name variable

print(name)
````

#### 2. Integers

Restrictions:
- **Whole numbers**

````python
# Let's create a variable called age and assign it to your age

age = 25 # Replace 25 with your actual age
print(age)
````

````python
# Let's combine those variables into one sentence using f-strings
# NOTE: f-strings are just an "f" in the beginning of the string and then you can use curly braces to insert variables into the string.


sentence = f"My name is {name} and I am {age} years old."
print(sentence)
````

#### 3. Float
Restrictions:
- **Decimal Places** 

````python
# Let's create a variable called height and assign it to your height in centimeters

height = 183.09 # Replace 183.09 with your actual height
print(height)
````

````python
# Let's combine those variables into one sentence using f-strings


sentence = f"My name is {name} and I am {age} years old and my height is {height} centimeters."
print(sentence)
````

#### 4. List

Restrictions:
- **Square Brackets "[ ]"**
- **The elements in a list are seperated by comma ","**

````python
# Let's create a list of your favorite Sri Lankan meals

favorite_meals = ["Meal1", "Meal2", "Meal3"] 
print(favorite_meals)
````

````python
# Some additional features of a list


## Lists can be indexed, which means you can access specific elements based on position. 
## First element is index 0.

print(favorite_meals[0])  # Prints the first meal in the list
````

````python
print(favorite_meals[1])  # Prints the second meal in the list
````

````python
## Lists can be modified, which means you can change the value of specific elements based on position.

favorite_meals[0] = "Sushi"  # Changes the first meal in the list
print(favorite_meals)  # Prints the modified list
````

````python
## Lists can be sliced, which means you can access a range of elements based on position.

print(favorite_meals[1:3])  # Prints the second and third meals in the list
````

````python
## If you want to remove an element by value, you can use the remove() method.

favorite_meals.pop(1) # Removes the second meal in the list
print(favorite_meals)  
````

````python
## If you want to remove a specific element by position, you can use the pop() method.

favorite_meals.remove("Sushi")  # Removes "Sushi" from the list
print(favorite_meals)  
````

````python
# Let's combine those variables into one sentence using f-strings
# The \n is a newline character that will create a new line.

sentence = f"My name is {name} and I am {age} years old and my height is {height} centimeters. \nMy favorite Sri Lankan meals are {favorite_meals}."
print(sentence)
````

#### 5. Tuples

Restrictions:
- **Curved Brackets "( )"**
- **The elements in a tuple are seperated by comma ","**

````python
# Unlike other data types, tuples are unchangeable.

thistuple = ("apple", "banana", "cherry")
print(thistuple)
````

````python
# Similar to a list you can index and slice a tuple, but you cannot modify it.

print(thistuple[1])
print(thistuple[0:2])
````

#### 6. Dictionary

Restrictions:
- **Curly Brackets "{ }"**
- **The elements in a dictionary are seperated by comma ","**

````python
# Lets put all the above information into a dictionary called my_info

my_info = {
    "name": name,
    "age": age,
    "height": height,
    "favorite_meals": favorite_meals
}

print(my_info)
````

````python
# Some additional features of a dictionary

## Dictionaries can be accessed using keys, which means you can access specific values based on the key.

print(my_info["name"]) 
````

````python
## To see all the keys in a dictionary, you can use the keys() method.

my_info.keys()
````

````python
## You can also see all the key-item pairs in a dictionary using the items() method.

my_info.items()
````

````python
## Dictionaries can also be modified, which means you can change the value of specific keys.

my_info["age"] = 99  # Changes the value associated with the "age" key
print(my_info)  
````

````python
## To create a new key-value pair in a dictionary, you can simply assign a value to a new key.

my_info["email"] = "random@gmail.com"  # Creates a new key-value pair
print(my_info)  
````

````python
## To delete a key-value pair in a dictionary, you can use the del statement.

del my_info["email"]  # Deletes the key-value pair associated with the "email" key
print(my_info)
````

### Control Statements

There are about **2** main types of control statements you need to know.

Control Statements help us control the flow and logic of the code. In otherwords it allows us to program a flow.


1. If-Else Statements - Ex: 
    ```python 
    if (a > b):
        do something
    else:
        do something
    ```

5. For Loops - Ex: 

    ```python 
    for item in items:
        do something
    ```

#### 1. If-Else Statements

Restrictions:
 - **Has to start with an "if"**
 - **Followed with a condition**
 - **: at the end**

````python
# Let's create a simple if-else statement that checks if you can drink.
## NOTE: We have already set our age variable earlier

legal_drinking_age = 21

if age >= legal_drinking_age:
    print("You are old enough to drink")
else:
    print("You are not old enough to drink")
````

````python
# We can also use elif statements to check for multiple conditions.

if age >= legal_drinking_age:
    print("Can Vote Can Drink")
elif age >= 18:
    print("Can Vote Can't Drink")
else:
    print("Can't Vote Can't Drink")
````

#### 1. For Loops

Restrictions:
 - **Has to start with an "for"**
 - **: at the end**

````python
# Let's create a new list of our favorite meals and print each meal using a for loop.

meals = ["Rice and Curry", "Hoppers", "Kottu", "String Hoppers", "Lamprais"]

for meal in meals:
    print(meal)
````

````python
# You can also use a for loop to iterate through a dictionary and print each key-value pair.

for key, value in my_info.items():
    print(f"{key}: {value}")
````

### Functions

Functions help us create a resuable piece of code which usually has a specific function.

```python 
def name_of_function(a,b):
  c = a + b

return c
```

Restrictions:

- **They need to start with "def"**
- **Must be named**


````python
# Let's create a function that takes in a name and age and returns a sentence using f-strings.

def create_sentence(name, age):
    return f"My name is {name} and I am {age} years old."
````

````python
create_sentence("Samantha", 32)
````

````python
# Notice how the function just does the work faster?
# You can reuse any code with a function so you avoid repeating yourself. 
# This is called DRY (Don't Repeat Yourself) principle in programming.

create_sentence("Gerad", 32)
````
