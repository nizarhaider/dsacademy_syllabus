### How to create a DataFrame?

You can either:
- Create it with the `pd.DataFrame()` function

    ```sql
    df = pd.DataFrame(
        data={}
    )
    ```


- Import a tabular dataset like a spreadsheet (csv/excel) or a parquet using `pd.import_csv()`
    ```sql
    df = pd.import_csv(
        path=""
    )
    ```


Let's look at how to create it with the `pd.DataFrame()` function first

````python
# We will create a DataFrame of names and ages using a dictionary.
## NOTE: The keys of the dictionary will be the column names and the values will be the data for each column.

dictionary_data = {
    "name": ["Nathushan", "Jerry", "Sameer", "Adil"],
    "age": [18, 25, 35, 30]
}

print(dictionary_data)
````

````python
# Now we pass the dict object as input to the data parameter of the DataFrame function and assign it to df variable.

df = pd.DataFrame(
    data=dictionary_data
)

df # NOTE: No need to use print() to display a dataframe in jupyter notebook.
````

### Missing Values

Almost always you will have missing data in the industry. 

Let's look into how to identify and manage them.

- To **identify** we can reuse the `df.info()` function
- To **drop** those values can use the `df.dropna()`
- To **fill** those values we can use `df.fillna()`


````python
# Like before we can use the .info() function to help with finding null or missing values

df.info()
````

````python
# From the above info we can see that vehicle column has only 8 non-null values 
# which means the rest is null 

# NOTE: NaN is the missing/empty values here

df['vehicle']
````

Now we can either drop the rows or the column of the null values.

Here is what it looks like with NaN rows dropped.

````python
df.dropna(
    axis=0  # 0 is row and 0 is column
)
````

Here is what it looks like with the NaN column ('Vehicle') dropped 

````python
df.dropna(
    axis=1
)
````

Now instead of dropping we can also fill the missing (NaN) values using `.fillna()` 

````python
# First let's check how many times the vehicles have been mentioned using the value_counts() function.

df['vehicle'].value_counts()
````

````python
# Then we get the mode() of the column which is ranked by most common to least

df['vehicle'].mode()
````

````python
# Then we use the most common vehicle to replace the NaN values.

df.fillna(
    value= df['vehicle'].mode()[0] # we need to pass a value to replace the NaN with.
)
````
